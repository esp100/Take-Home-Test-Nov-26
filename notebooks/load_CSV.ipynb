```python
FECHA_CORRIDA = pd.Timestamp.today().normalize()
```

```python
df_dim_clientes = df_raw_clientes.copy()
df_dim_clientes["region"] = (
    df_dim_clientes["region"]
    .str.strip()
    .str.title()
    .fillna("No Especificada")
)
df_dim_clientes["channel"] = df_dim_clientes["channel"].str.strip()
df_dim_clientes["signup_date"] = pd.to_datetime(
    df_dim_clientes["signup_date"], format="%Y-%m-%d", errors="coerce"
)
df_dim_clientes = df_dim_clientes[
    ["customer_id", "customer_name", "region", "channel", "signup_date"]
].reset_index(drop=True)

df_dim_clientes.head()
```

```python
df_dim_productos = df_raw_productos.copy()
df_dim_productos["category"] = df_dim_productos["category"].str.strip()
df_dim_productos["unit_price"] = pd.to_numeric(
    df_dim_productos["unit_price"], errors="coerce"
)
df_dim_productos["is_active"] = df_dim_productos["is_active"].astype(bool)
df_dim_productos = df_dim_productos[
    ["product_id", "product_name", "category", "unit_price", "is_active"]
].reset_index(drop=True)

df_dim_productos.head()
```

```python
# 3.1 Reducir por order_id (duplicados idénticos en todas las columnas)
df_pedidos_base = (
    df_raw_pedidos
    .drop_duplicates(subset=["order_id"], keep="first")
    .reset_index(drop=True)
)

# 3.2 Parseo de fechas: YYYY-MM-DD y fallback dd/mm/yyyy (locale es-MX)
order_ymd = pd.to_datetime(df_pedidos_base["order_date"], format="%Y-%m-%d", errors="coerce")
order_dmy = pd.to_datetime(df_pedidos_base["order_date"], format="%d/%m/%Y", errors="coerce")
df_pedidos_base["order_date_parsed"] = order_ymd.fillna(order_dmy)

df_pedidos_base["cancel_date_parsed"] = pd.to_datetime(
    df_pedidos_base["cancel_date"], format="%Y-%m-%d", errors="coerce"
)

# 3.3 Split clean vs quarantine por validez de customer_id
mask_customer_ok = (
    df_pedidos_base["customer_id"].notna()
    & df_pedidos_base["customer_id"].isin(df_dim_clientes["customer_id"])
)

df_fact_pedidos = df_pedidos_base[mask_customer_ok].copy()
df_quar_pedidos = df_pedidos_base[~mask_customer_ok].copy()

df_quar_pedidos["motivo"] = np.where(
    df_quar_pedidos["customer_id"].isna(),
    "customer_id_nulo",
    "customer_id_huerfano",
)
df_quar_pedidos["fecha_deteccion"] = FECHA_CORRIDA
df_quar_pedidos = df_quar_pedidos.rename(columns={
    "order_date": "order_date_raw",
    "cancel_date": "cancel_date_raw",
    "order_date_parsed": "order_date",
    "cancel_date_parsed": "cancel_date",
})[
    ["order_id", "customer_id", "order_date", "status", "order_total",
     "cancel_date", "order_date_raw", "cancel_date_raw", "motivo", "fecha_deteccion"]
].reset_index(drop=True)

df_fact_pedidos.head()
```

```python
df_lineas_base = df_raw_lineas_pedido.copy()
calc_amount = (
    df_lineas_base["quantity"]
    * df_lineas_base["unit_price"]
    * (1 - df_lineas_base["discount_pct"])
).round(2)

mask_bad_qty       = df_lineas_base["quantity"] <= 0
mask_orphan_prod   = ~df_lineas_base["product_id"].isin(df_dim_productos["product_id"])
mask_bad_amount    = ~np.isclose(
    df_lineas_base["line_amount"], calc_amount, rtol=0, atol=0.005
)
mask_bad_line = mask_bad_qty | mask_orphan_prod | mask_bad_amount

df_quar_lineas_self = df_lineas_base[mask_bad_line].copy()
df_quar_lineas_self["motivo"] = np.select(
    [mask_bad_qty[mask_bad_line],
     mask_orphan_prod[mask_bad_line],
     mask_bad_amount[mask_bad_line]],
    ["cantidad_no_positiva", "producto_huerfano", "importe_no_cuadra"],
    default="otro",
)
df_quar_lineas_self["fecha_deteccion"] = FECHA_CORRIDA

df_lineas_ok = df_lineas_base[~mask_bad_line].copy()

# Cascade: si el pedido está en cuarentena, sus líneas también
mask_order_quar = df_lineas_ok["order_id"].isin(df_quar_pedidos["order_id"])
df_quar_lineas_cascade = df_lineas_ok[mask_order_quar].copy()
df_quar_lineas_cascade["motivo"] = "pedido_en_cuarentena"
df_quar_lineas_cascade["fecha_deteccion"] = FECHA_CORRIDA

df_fact_lineas = df_lineas_ok[~mask_order_quar].copy()

df_quar_lineas_pedido = pd.concat(
    [df_quar_lineas_self, df_quar_lineas_cascade], ignore_index=True
)[["line_id", "order_id", "product_id", "quantity", "unit_price",
   "discount_pct", "line_amount", "motivo", "fecha_deteccion"]]

df_fact_lineas = df_fact_lineas[
    ["line_id", "order_id", "product_id", "quantity", "unit_price",
     "discount_pct", "line_amount"]
].reset_index(drop=True)

df_quar_lineas_pedido.head()
```

```python
import os
from fastavro import writer as avro_writer, parse_schema

OUTPUT_DIR = "../data/processed/"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ---------- Helpers de exportación ----------
def save_parquet(df, nombre):
    path = os.path.join(OUTPUT_DIR, f"{nombre}.parquet")
    df.to_parquet(path, index=False, engine="pyarrow", compression="snappy")
    print(f"  [PARQUET] {path}  ({os.path.getsize(path)/1024:.1f} KB)")
    return path

def _to_avro_records(df):
    """Convierte un DataFrame a lista de dicts compatible con fastavro."""
    df_out = df.copy()
    for col in df_out.columns:
        if pd.api.types.is_datetime64_any_dtype(df_out[col]):
            df_out[col] = df_out[col].dt.strftime("%Y-%m-%d")
    records = df_out.astype(object).where(pd.notna(df_out), None).to_dict("records")
    for rec in records:
        for k, v in rec.items():
            if isinstance(v, np.integer):
                rec[k] = int(v)
            elif isinstance(v, np.floating):
                rec[k] = float(v)
            elif isinstance(v, np.bool_):
                rec[k] = bool(v)
    return records

def save_avro(df, nombre, schema_dict):
    path = os.path.join(OUTPUT_DIR, f"{nombre}.avro")
    parsed = parse_schema(schema_dict)
    with open(path, "wb") as f:
        avro_writer(f, parsed, _to_avro_records(df))
    print(f"  [AVRO]    {path}  ({os.path.getsize(path)/1024:.1f} KB)")
    return path

# ---------- Esquemas Avro ----------
def _rec(name, fields):
    return {"type": "record", "name": name, "namespace": "dwh", "fields": fields}

AVRO_DIM_CLIENTES = _rec("dim_clientes", [
    {"name": "customer_id",   "type": "string"},
    {"name": "customer_name", "type": "string"},
    {"name": "region",        "type": "string"},
    {"name": "channel",       "type": "string"},
    {"name": "signup_date",   "type": ["null", "string"], "default": None},
])

AVRO_DIM_PRODUCTOS = _rec("dim_productos", [
    {"name": "product_id",   "type": "string"},
    {"name": "product_name", "type": "string"},
    {"name": "category",     "type": "string"},
    {"name": "unit_price",   "type": "double"},
    {"name": "is_active",    "type": "boolean"},
])

AVRO_FACT_PEDIDOS = _rec("fact_pedidos", [
    {"name": "order_id",           "type": "string"},
    {"name": "customer_id",        "type": "string"},
    {"name": "order_date",         "type": ["null", "string"], "default": None},
    {"name": "status",             "type": "string"},
    {"name": "order_total",        "type": "double"},
    {"name": "cancel_date",        "type": ["null", "string"], "default": None},
    {"name": "total_from_lines",   "type": "double"},
    {"name": "diff_total",         "type": "double"},
    {"name": "lines_match_total",  "type": "boolean"},
])

AVRO_FACT_LINEAS = _rec("fact_lineas_pedido", [
    {"name": "line_id",       "type": "string"},
    {"name": "order_id",      "type": "string"},
    {"name": "product_id",    "type": "string"},
    {"name": "quantity",      "type": "int"},
    {"name": "unit_price",    "type": "double"},
    {"name": "discount_pct",  "type": "double"},
    {"name": "line_amount",   "type": "double"},
])

AVRO_QUAR_PEDIDOS = _rec("quar_pedidos", [
    {"name": "order_id",         "type": "string"},
    {"name": "customer_id",      "type": ["null", "string"], "default": None},
    {"name": "order_date",       "type": ["null", "string"], "default": None},
    {"name": "status",           "type": "string"},
    {"name": "order_total",      "type": "double"},
    {"name": "cancel_date",      "type": ["null", "string"], "default": None},
    {"name": "order_date_raw",   "type": ["null", "string"], "default": None},
    {"name": "cancel_date_raw",  "type": ["null", "string"], "default": None},
    {"name": "motivo",           "type": "string"},
    {"name": "fecha_deteccion",  "type": ["null", "string"], "default": None},
])

AVRO_QUAR_LINEAS = _rec("quar_lineas_pedido", [
    {"name": "line_id",         "type": "string"},
    {"name": "order_id",        "type": "string"},
    {"name": "product_id",      "type": "string"},
    {"name": "quantity",        "type": "int"},
    {"name": "unit_price",      "type": "double"},
    {"name": "discount_pct",    "type": "double"},
    {"name": "line_amount",     "type": "double"},
    {"name": "motivo",          "type": "string"},
    {"name": "fecha_deteccion", "type": ["null", "string"], "default": None},
])
```
